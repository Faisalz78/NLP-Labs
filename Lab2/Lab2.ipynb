# Task 1: Hashtag Extraction and Analysis

In this task, the Apple Twitter dataset is analyzed using regular expressions.

The objectives are to:

- Extract all hashtags from the tweets.
- Calculate the total number of hashtags.
- Calculate the number of unique hashtags.
- Identify the top 10 most frequently used hashtags.

The regular expression `#\w+` is used to detect hashtags.

- `#` matches the hashtag symbol.
- `\w` matches letters, digits, and underscores.
- `+` means one or more characters.

The tweets are converted to lowercase so that hashtags such as `#Apple` and `#apple` are treated as the same hashtag.

The `Counter` class is then used to calculate the frequency of each hashtag and display the top 10 most frequently used hashtags.

```python
# ============================================================
# Task 1: Load Apple Twitter Dataset
# ============================================================

DATASET_PATH = "apple-twitter-sentiment-texts.csv"

df = pd.read_csv(DATASET_PATH)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python
# ============================================================
# Task 1: Identify Tweet Text Column
# ============================================================

possible_columns = [
    "text",
    "tweet",
    "tweets",
    "tweet_text",
    "sentiment_text"
]

text_column = None

for column in possible_columns:
    if column in df.columns:
        text_column = column
        break

if text_column is None:
    raise ValueError(
        f"Tweet text column not found. Available columns: {df.columns.tolist()}"
    )

print("Tweet text column:", text_column)
```

```python
# ============================================================
# Task 1: Extract Hashtags Using Regular Expressions
# ============================================================

hashtag_pattern = r"#\w+"

all_hashtags = []

for tweet in df[text_column].dropna().astype(str):

    # Convert text to lowercase to avoid counting
    # #Apple and #apple as different hashtags
    tweet = tweet.lower()

    hashtags = re.findall(hashtag_pattern, tweet)

    all_hashtags.extend(hashtags)


print("Total number of hashtags:", len(all_hashtags))
print("Number of unique hashtags:", len(set(all_hashtags)))
```

```python
# ============================================================
# Task 1: Top 10 Most Frequently Used Hashtags
# ============================================================

hashtag_counts = Counter(all_hashtags)

top_10_hashtags = hashtag_counts.most_common(10)

top_10_df = pd.DataFrame(
    top_10_hashtags,
    columns=["Hashtag", "Frequency"]
)

print("Top 10 Most Frequently Used Hashtags:")
display(top_10_df)
```

# Task 2: Using `re.compile()`

In this task, `re.compile()` is used to create a reusable regular expression pattern.

The pattern `\d+` matches one or more digits.

The compiled pattern is used to replace the year `2021` with `2022`.

`re.compile()` is useful because it converts a regular expression into a reusable pattern object. This makes the code cleaner and more efficient when the same pattern is used multiple times.

```python
# ============================================================
# Task 2: Using re.compile()
# ============================================================

text = "This year is 2021"

# Create a compiled regex pattern that matches one or more digits
pattern = re.compile(r"\d+")

# Print the type of the compiled pattern
print("Pattern type:", type(pattern))

# Replace 2021 with 2022
updated_text = pattern.sub("2022", text)

print("Original text:", text)
print("Updated text :", updated_text)
```

# Task 3: Using `re.split()`

In this task, `re.split()` is used to split a string based on a regular expression pattern.

The pattern `\d+` matches one or more digits.

Therefore, the numbers `11`, `2`, `3`, and `4` are used as splitting points in the text.

`re.split()` is useful when we want to divide text using a pattern instead of a single fixed character.

```python
# ============================================================
# Task 3: Using re.split()
# ============================================================

text = "a 11 b 2 3 c 4"

# Split the text whenever one or more digits appear
result = re.split(r"\d+", text)

print("Original text:", text)
print("Split result :", result)
```

# Task 4: Tokenization Using spaCy and NLTK

In this task, the sentence `"I'm enjoying the NLP course!"` is tokenized using both spaCy and NLTK.

The function `spacy.load("en_core_web_sm")` loads the small English language model provided by spaCy.

This model provides several NLP features, including:

- Tokenization
- Part-of-speech tagging
- Lemmatization
- Named Entity Recognition

The same sentence is also tokenized using `nltk.word_tokenize()`.

For this sentence, spaCy and NLTK produce the same tokens. However, both libraries use different tokenization rules, so their outputs may differ when processing more complex text.

```python
# ============================================================
# Task 4: Tokenization Using spaCy
# ============================================================

# Load the small English model
nlp = spacy.load("en_core_web_sm")

text = "I'm enjoying the NLP course!"

# Process the sentence
doc = nlp(text)

print("spaCy Tokens:")
print("-" * 20)

for token in doc:
    print(token.text)
```

```python
# ============================================================
# Task 4: Tokenization Using NLTK
# ============================================================

text = "I'm enjoying the NLP course!"

nltk_tokens = nltk.word_tokenize(text)

print("NLTK Tokens:")
print("-" * 20)

for token in nltk_tokens:
    print(token)
```

```python
# ============================================================
# Task 4: Compare spaCy and NLTK Tokenization
# ============================================================

spacy_tokens = [token.text for token in nlp(text)]
nltk_tokens = nltk.word_tokenize(text)

print("spaCy Tokens:")
print(spacy_tokens)

print("\nNLTK Tokens:")
print(nltk_tokens)

print("\nNumber of spaCy tokens:", len(spacy_tokens))
print("Number of NLTK tokens :", len(nltk_tokens))
```
